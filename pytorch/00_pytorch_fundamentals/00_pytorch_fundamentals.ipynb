# 如果是高维矩阵
以下两个tensor A, B:
1. A.shape = [p, q, r]
2. B.shape = [s, t, r, v]

最后两维单独拿出来做矩阵乘法: [q, r] @ [r, v] = [q, v]

前面所有维度全部视为批量 batch 维度, 每个维度进行广播处理:
1. 对应相等, 合法, 不变
2. 其中一个是 1 -> 自动复制扩张匹配另一个
3. 都不为1且不相等 -> 失败

eg.

A [1, 4, 5] @ B [3, 5, 2] = [3, 4, 2]

A [2, 1, 5, 5] @ [1, 7, 5, 3] = [2, 7, 5, 3]

A [2, 1, 5, 5, 8, 9, 10] @ [1, 7, 5, 1, 1, 10, 4] = [2, 7, 5, 5, 8, 9, 4]

```python
t1 = torch.rand(1, 2, 1, 3, 6)
t2 = torch.rand(6, 1, 5, 6, 7)
t3 = t1 @ t2
print(f"t1 shape: {t1.shape} dot t2 shape: {t2.shape}")
print("equals:")
print(f"t1 @ t2 shape: {(t1 @ t2).shape}")
```

# Reshape、 View、 Stack、 Squeeze、Unsqueeze

```python
# View
tensor = torch.arange(1, 9)
tensor_view = tensor.view(1, 8)
print(f"{tensor} shape: {tensor.shape}")
print(f"{tensor_view} shape: {tensor_view.shape}")
```

```python
# View共享内存, 修改一个会修改另一个
tensor_view[:, 2] += 10
print(f"{tensor} shape: {tensor.shape}")
print(f"{tensor_view} shape: {tensor_view.shape}")
```

```python
# View要求张量连续(经过transpose/permute之后就不连续了, 需要.contiguous)
tensor = torch.tensor([range(1, 9), range(11, 19)])
tensor_t = tensor.T
print(f"original: {tensor} \nshape: {tensor.shape}\n")
print(f"transpose: {tensor_t} \nshape: {tensor_t.shape}\n")

tt_view = tensor_t.contiguous().view(1, 16)
print(f"view: {tt_view} \nshape: {tt_view.shape}\n")
```

```python
# reshape, 内存连续时共享内存, 等同于view
tensor = torch.arange(1, 10)
re_tensor = tensor.reshape(3, 3)
print(f"{tensor} \nreshaped as : \n{re_tensor}\n")

re_tensor[2, 2] = 99
print(f"after modify: \n{tensor} \nreshaped as : \n{re_tensor}\n")
```

```python
# 如果是内存不连续tensor(如转置)的进行reshape, 不会共享内存, 会新开辟
re_tt = torch.ones(3, 3).T
rere_tt = re_tt.reshape(1, 9)
print(f"before modify: \n{re_tt} \nreshaped as : \n{rere_tt}\n")

rere_tt[0, 3] = 66
print(f"after modify: \n{re_tt} \nreshaped as : \n{rere_tt}\n")
```

```python
# stack 合并tensor, 插入一个全新维度. dim指的是新增的维度的位置
# dim = 0 相当于上下叠起来
# dim = 1 相当于把两个tensor同一个位置的叠起来

# 设单个张量 shape = [D0, D1, D2]，有 N 个张量堆叠：
# dim=0 → 新 shape：[N, D0, D1, D2]
# dim=1 → 新 shape：[D0, N, D1, D2]
# dim=2 → 新 shape：[D0, D1, N, D2]

t1 = torch.tensor([[1, 2, 3, 4],
                  [5, 6, 7, 8]])
t2 = torch.tensor([[9, 10, 11, 12],
                  [13, 14, 15, 16]])
t3 = torch.tensor([[17, 18, 19, 20],
                  [21, 22, 23, 24]])
t4 = torch.stack([t1, t2, t3], dim=0)
t5 = torch.stack([t1, t2, t3], dim=1)
print(f"{t4} \nshape: \n{t4.shape}\n")
print(f"{t5} \nshape: \n{t5.shape}\n")
```

```python
# squeeze: 把所有shape为1的维度去掉
tensor = torch.rand(1, 1, 2, 3)
print(f"previous shape: {tensor.shape}, after shape: {tensor.squeeze().shape}")
```

```python
# unsqueeze: 在指定dim新增一个维度
tensor = torch.rand(3, 4, 5)
print(f"previous shape: {tensor.shape}, after shape:")
print(f"dim = 0 -> {tensor.unsqueeze(0).shape}")
print(f"dim = 1 -> {tensor.unsqueeze(1).shape}")
print(f"dim = 2 -> {tensor.unsqueeze(2).shape}")
print(f"dim = 3 -> {tensor.unsqueeze(3).shape}")
```

```python
# permute - 排列组合  将不同维度重新排列 permute(a, b, c): idx=0用idx=a的, idx=1的用idx=b的, ...
# return 一个 view, 共享内存
# 常用于图片 height weight color_channels
tensor = torch.rand(224, 224, 3)
print(f"before: \n{tensor.shape}\n")
print(f"after(color to first): \n{tensor.permute(2, 0, 1).shape}\n")
```

# PyTorch and NumPy

经常拿到NumPy格式的数据, 但是想要使用的是PyTorch的Tensor. 
- NumPy -> Tensor 可以使用`torch.from_numpy(ndarray)`来转换
- Tensor -> NumPy 可以使用`torch.Tensor.numpy()`来转换

注意, 这种转换是复制一份, 不是共享空间

```python
import numpy as np
```

```python
# numpy -> tensor
# numpy默认浮点类型是float64, 因此转换为tensor也是float64
arr = np.arange(1.0, 10.0)
tensor = torch.from_numpy(arr)
arr, arr.dtype, tensor, tensor.dtype
```

```python
# tensor -> numpy
# tensor默认浮点类型是float32, 因此转换为numpy也是float32
tensor = torch.arange(1.0, 10.0)
arr = torch.Tensor.numpy(tensor)
arr, arr.dtype, tensor, tensor.dtype
```

# Reproducibility

```python
# 设置了RANDOM_SEED, 每次都会得到相同的结果
RANDOM_SEED = 42
torch.manual_seed(RANDOM_SEED)
t_A = torch.rand(4, 2)
t_B = torch.rand(4, 2)
t_A, t_B
```

```python
# RANDOM_SEED只生效一次, 如果需要重置就要再次执行
RANDOM_SEED = 42
torch.manual_seed(RANDOM_SEED)
# 该t_A与上面cell里的相同
t_A = torch.rand(4, 2)
t_A
```

```python
# 如果使用了PyTorch以外的随机, 也需要设置对应的随机种子
# Python:
import random
random.seed(42)
# NumPy:
np.random.seed(42)
```
